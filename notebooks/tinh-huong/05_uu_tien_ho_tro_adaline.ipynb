## Bối cảnh

Một nhóm hỗ trợ muốn đưa các phiếu khẩn cấp lên đầu hàng đợi.

Feature có thể cân nhắc: số từ khẩn cấp, dịch vụ có bị gián đoạn hay không, thời gian chờ, mức khách hàng (mã hoá số).

## Vì sao dùng Adaline thay Perceptron?

Thay vì chỉ 0/1 (thông thường / ưu tiên), hãy coi target là **điểm mức độ ưu tiên** trong [0, 1] hoặc [0, 10], để có thể xếp hạng toàn bộ hàng đợi, không chỉ chia hai nhóm.

Nhắc lại công thức Adaline / Widrow–Hoff:

$$
\text{net} = \mathbf{w}^\top \mathbf{x} + b, \qquad e = t - \text{net}
$$

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta e \mathbf{x}, \qquad b \leftarrow b + \eta e
$$

Điểm ưu tiên liên tục cho phép sắp xếp toàn bộ hàng đợi theo thứ tự, việc mà Perceptron 0/1 không làm được trực tiếp.
