## Vì sao dùng Adaline thay Perceptron?

Thay vì chỉ gán nhãn 0/1 (máy bình thường / cần dừng), hãy nghĩ target là **điểm nguy cơ quá tải** trong khoảng [0, 1]: 0 nghĩa là hoàn toàn an toàn, 1 nghĩa là chắc chắn cần dừng máy. Nhiều máy sẽ có điểm trung gian (0.3, 0.6...).

Nhắc lại công thức Adaline / Widrow–Hoff:

$$
\text{net} = \mathbf{w}^\top \mathbf{x} + b, \qquad e = t - \text{net}
$$

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta e \mathbf{x}, \qquad b \leftarrow b + \eta e
$$

Vì bỏ sót một máy quá tải nguy hiểm hơn báo động nhầm, hãy thử đặt ngưỡng cảnh báo thấp hơn 0.5 và quan sát ảnh hưởng lên số lượng cảnh báo giả.
