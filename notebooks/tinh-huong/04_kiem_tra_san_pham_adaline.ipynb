## Bối cảnh

Một dây chuyền cần phân loại nhanh sản phẩm đạt hay cần chuyển sang kiểm tra lại.

Feature có thể cân nhắc: độ lệch khối lượng, độ lệch kích thước, số vết xước, điểm màu sắc (dùng độ lệch tuyệt đối).

## Vì sao dùng Adaline thay Perceptron?

Thay vì chỉ 0/1 (đạt / cần kiểm tra lại), hãy coi target là **điểm lỗi tổng hợp** trong [0, 1], phản ánh mức độ sản phẩm lệch khỏi tiêu chuẩn.

Nhắc lại công thức Adaline / Widrow–Hoff:

$$
\text{net} = \mathbf{w}^\top \mathbf{x} + b, \qquad e = t - \text{net}
$$

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta e \mathbf{x}, \qquad b \leftarrow b + \eta e
$$

Điểm lỗi tổng hợp liên tục giúp xếp hạng sản phẩm theo mức độ nghiêm trọng, không chỉ chia hai nhóm như Perceptron.
