## Vì sao dùng Adaline thay Perceptron?

Thay vì chỉ 0/1 (theo dõi thường / cần hỗ trợ), hãy coi target là **điểm mức độ cần hỗ trợ** trong [0, 1], để giảng viên có thể sắp xếp thứ tự ưu tiên gặp sinh viên.

Nhắc lại công thức Adaline / Widrow–Hoff:

$$
\text{net} = \mathbf{w}^\top \mathbf{x} + b, \qquad e = t - \text{net}
$$

$$
\mathbf{w} \leftarrow \mathbf{w} + \eta e \mathbf{x}, \qquad b \leftarrow b + \eta e
$$

Điểm mức độ cần hỗ trợ liên tục giúp giảng viên sắp xếp thứ tự ưu tiên gặp từng sinh viên, thay vì chỉ chia hai nhóm như Perceptron.
